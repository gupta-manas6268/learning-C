## Calculate a weighted rating

```python
# Formula from 'imdb' website.
#       WR = ((v / (v+m)) * R) + ((m / (v+m)) * C)
# 
# (WR- Weighted Rating)
# (v- vote count)
# (m- min. no. of votes required (to get in Recommendation-system))
# (R- (i.e. vote avg.) => avg.(average) rating of the movie) 
# (C- avg. rating across all movies)
```

```python
m = movies["vote_count"].quantile(0.9)
# .quantile(0.9) => i.e. 90% of movies have ("vote_count" < 1838)
#            So, we are taking only top-10% of movies in our model.
m
```

```python
movies_filtered = movies.copy().loc[movies["vote_count"] >= m]
movies_filtered
```

```python
C = movies["vote_average"].mean()
C
```

```python
def weighted_rating(df, m=m, C=C): # df => data-frame
    R = df["vote_average"]
    v = df["vote_count"]
    wr = ((v / (v+m)) * R) + ((m / (v+m)) * C)
    
    return wr
```

```python
#     New Column adding. (↓)
movies_filtered["weighted_rating"] =      movies_filtered.apply(weighted_rating, axis=1)
# (It takes 'df' => movies_filtered, because it '.apply()' function.) (↑)
movies_filtered 
```
