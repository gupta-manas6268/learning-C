# Grouping

```python
# SQL code.
''' 
    SELECT
        transmission_type,
        AVG(MSRP)
    FROM
        cars
    GROUP BY
        transmission_type
'''
```

```python
df.groupby('Transmission Type').MSRP.mean() 
# .groupby => Groups values like 'AUTOMATIC', etc. in same values.
```

```python
df.groupby('Transmission Type').MSRP.max()
```
